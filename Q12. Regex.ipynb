```python
from pyspark.sql.functions import regexp_replace, col, replace, regexp_extract
```

```python
data=[(1,"Sagar-Prajapati"),(2,"Alex-John"),(3,"John Cena"),(4,"Kim Joe")]
schema="ID int,Name string"
df=spark.createDataFrame(data,schema)
```

```python
df.display()
```

```python
df1= df.withColumn("new", regexp_replace(col("Name"),r'\-',r' ' ))
df1.display()
```

```python
df2  =  df.withColumn("fname", regexp_extract(col("Name"),r'([A-Za-z]+)\-([A-Za-z]+)',1))\
        .withColumn("lname", regexp_extract(col("Name"),r'([A-Za-z]+)\-([A-Za-z]+)',2))
df2.display()
```
