```python
from pyspark.sql.functions import *
```

```python
data=[('Genece' , 2 , 75000),
('𝗝𝗮𝗶𝗺𝗶𝗻' , 2 , 80000 ),
('𝗣𝗮𝗻𝗸𝗮𝗷' , 2 , 80000 ),
('Tarvares' , 2 , 70000),
('Marlania' , 4 , 70000),
('Briana' , 4 , 85000),
('𝗞𝗶𝗺𝗯𝗲𝗿𝗹𝗶' , 4 , 55000),
('𝗚𝗮𝗯𝗿𝗶𝗲𝗹𝗹𝗮' , 4 , 55000),  
('Lakken', 5, 60000),
('Latoynia' , 5 , 65000) ]
schema="emp_name string,dept_id int,salary int"
df=spark.createDataFrame(data,schema)
```

```python
df.display()
```

```python
from pyspark.sql.functions import max, min
df1= df.groupby("dept_id").agg(max("salary").alias("mx"), min("salary").alias("mn"))

df1.display()
```

```python
df2= df.join(df1, df["dept_id"]==df1["dept_id"], "inner").select(df["*"], df1["mx"], df1["mn"])
df2.display()
```

```python
df3 =  df2.select("emp_name", "dept_id", when(col("salary") == col("mx"), "max_salary").when(col("salary") == col("mn"), "mn_salary").alias("flag"))

df3.display()
```

```python
df4= df3.filter("flag == 'max_salary'").groupby("dept_id").agg(collect_set("emp_name").alias("max_emp"),collect_set("emp_name").alias("min_emp"))
df4.display()
```
