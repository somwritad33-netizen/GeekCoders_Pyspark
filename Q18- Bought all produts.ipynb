```python
data=[(1,5),(2,6),(3,5),(3,6),(1,6)]
schema="customer_id int,product_key int"
customer_df=spark.createDataFrame(data,schema)


data=[(5,),(6,)]
schema="product_key int"
product_df=spark.createDataFrame(data,schema)
```

```python
customer_df.display()

customer_df1 =  customer_df.groupby("customer_id").agg(collect_set(col("product_key")).alias("prod"))
customer_df1.display()
```

```python
from pyspark.sql.functions import collect_set, col, lit
product_df.display()

product_df1 =  product_df.agg(collect_set(col("product_key")).alias("prods"))
prods = product_df1.collect()[0][0]
prods
```

```python
from pyspark.sql.functions import col, array
df = customer_df1.withColumn("flag", col("prod") == array(*[lit(i) for i in prods]))
df.display()
```
