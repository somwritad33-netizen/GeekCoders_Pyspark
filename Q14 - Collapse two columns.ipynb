```python
from pyspark.sql.functions import col, struct
```

```python
data = [
    ("john", "tomato", 2),
    ("𝚋𝚒𝚕𝚕", "𝚊𝚙𝚙𝚕𝚎", 2),
    ("john", "𝚋𝚊𝚗𝚊𝚗𝚊", 2),
    ("john", "tomato", 3),
    ("𝚋𝚒𝚕𝚕", "𝚝𝚊𝚌𝚘", 2),
    ("𝚋𝚒𝚕𝚕", "𝚊𝚙𝚙𝚕𝚎", 2),
]
schema = "name string,item string,weight int"
df = spark.createDataFrame(data, schema)
```

```python
df.display()
```

```python
df1 = df.groupBy("name", "item").count()
df1.display()
```

```python
df2 = df1.select("*").collect()
df2


l=([(row[0], (row[1], row[2])) for row in df2])
for i in l:
    print(i[0],i[1] )
```
