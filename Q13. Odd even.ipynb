```python
data=[(1, 'War','great 3D',8.9)
,(2, 'Science','fiction',8.5) 
,(3, 'irish','boring',6.2)
,(4, 'Ice song','Fantacy',8.6)  
,(5, 'House card','Interesting',9.1)]    
schema="ID int,movie string,description string,rating double"
df=spark.createDataFrame(data,schema)
```

```python
df.display()
```

```python
from pyspark.sql.functions import expr, col, when
```

```python
df1=df.select("*",when(df["ID"]%2!=0,df["ID"]).alias("new")).filter((col("new").isNotNull()) & (~col("description").like("%boring%")))
df1.display()
```

```python
df1=df.select("*",when(df["ID"]%2!=0,df["ID"]).alias("new")).filter('new IS NOT NULL AND description NOT LIKE "%boring%"')
df1.display()
```
