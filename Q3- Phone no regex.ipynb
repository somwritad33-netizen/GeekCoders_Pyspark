```python
import regex as re
from pyspark.sql.functions import count, aggregate, length
from pyspark.sql.functions import col, rlike
```

```python
df = spark.createDataFrame([(1, "a","U975869Y9i"),(2,"b","9108088303"),(3, "c",'z09855e5f'),(4, "d", '811D076328')],["id","name","ssn"])
display(df)
```

```python
from pyspark.sql.functions import length
df = df.withColumn("ssn_ct", length("ssn"))
df = df.filter((col("ssn_ct") >= 10) & (col("ssn").rlike("^[0-9]*$")))



df.display()
```
