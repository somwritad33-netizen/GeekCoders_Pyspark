```python
data = [(1, "abc@gmail.com"), (2, "bcd@gmail.com"), (3, "abc@gmail.com")]
schema = "ID int,email string"
df = spark.createDataFrame(data, schema)
df.display()
```

```python
from pyspark.sql.functions import col
df1 =  df.groupby("email").count().filter(col("count")>1)
df1.display()
```
